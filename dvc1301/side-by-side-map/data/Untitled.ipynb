```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import os
os.listdir()
```

```python
df = pd.read_csv('data.csv')
df.head()
```

```python
df['region'] = df.AREACD.str.contains('E09')
df[df.region]
```

```python
fig, ax = plt.subplots(figsize=(10,10))
ax.plot(df[df.region].lockdown_mobility, df[df.region].post_lockdown_mobility, '.')
ax.plot(df[~df.region].lockdown_mobility, df[~df.region].post_lockdown_mobility, '.')
ax.plot([-100,350],[-100,350])
ax.set_xlabel('% change in mobility in lockdown')
ax.set_ylabel('% change in mobility after lockdown')
ax.set_xlim([-100,350])
ax.set_ylim([-100,350])
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,10))
ax.plot(df[df.region].perc_flats, df[df.region].post_lockdown_mobility - df[df.region].lockdown_mobility, '.')
ax.plot(df[~df.region].perc_flats, df[~df.region].post_lockdown_mobility - df[~df.region].lockdown_mobility, '.')
ax.set_xlabel('% in flats')
ax.set_ylabel('increase in % change mobility from baseline between lockdown and after lockdown')
ax.set_xlim([-100,150])
ax.set_ylim([-30,350])
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,10))
ax.plot(df[df.region].perc_garden, df[df.region].post_lockdown_mobility - df[df.region].lockdown_mobility, '.')
ax.plot(df[~df.region].perc_garden, df[~df.region].post_lockdown_mobility - df[~df.region].lockdown_mobility, '.')
ax.set_xlabel('% properties that have gardens')
ax.set_ylabel('increase in % change mobility from baseline between lockdown and after lockdown')
# ax.set_xlim([-100,150])
# ax.set_ylim([-30,350])
plt.show()
```
